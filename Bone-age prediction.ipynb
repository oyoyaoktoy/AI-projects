### what this code does:

- **from sklearn.model_selection import train_test_split**
  This imports scikit-learn's function for randomly dividing a dataset into two parts. It's used twice below to create three parts.

- **train_split, temp_split = train_test_split(...)**
  This is the first split. It takes 60% of the data for training (`train_size=0.60`) and holds the remaining 40% in a temporary set (`temp_split`) to be divided again.
  - `stratify=train_df["age_bracket"]` keeps the proportion of each age bracket the same in both parts. If 10% of all images are in the 157-228 month bracket, roughly 10% of the training set and 10% of the temporary set will be too.
  - `random_state=42` fixes the random seed so you get the same split every time you run the code.

- **val_split, test_split = train_test_split(...)**
  This is the second split. It divides the temporary 40% in half (`train_size=0.50`), giving 20% for validation and 20% for testing. It is stratified by `age_bracket` again, so all three sets end up with the same age distribution. The final split is 60% train, 20% validation and 20% test.

- **print("Train:", ...), print("Validation:", ...), print("Test:", ...)**
  `.shape[0]` gives the number of rows in each set. Printing them lets you confirm the sizes match the intended 60/20/20 split.

- **train_ids, val_ids, test_ids = set(...)**
  This collects the unique `id` values from each split into Python sets. Sets make it easy to compare which images ended up in which split.

- **print("Train/Val overlap:", len(train_ids & val_ids))** (and the Train/Test and Val/Test lines)
  The `&` operator finds the IDs that appear in both sets, and `len()` counts them. This is a data leakage check. Every overlap should print `0`, meaning no image appears in more than one split. If the numbers were above zero, the model would be tested on images it had already seen in training, and its scores would look better than they should.

```python
from sklearn.model_selection import train_test_split

train_split, temp_split = train_test_split(
    train_df,
    train_size=0.60,
    stratify=train_df["age_bracket"],
    random_state=42 
)

val_split, test_split = train_test_split(
    temp_split,
    train_size=0.50,
    stratify=temp_split["age_bracket"],
    random_state=42
)

print("Train:", train_split.shape[0])
print("Validation:", val_split.shape[0])
print("Test:", test_split.shape[0])

train_ids, val_ids, test_ids = set(train_split["id"]), set(val_split["id"]), set(test_split["id"])
print("Train/Val overlap:", len(train_ids & val_ids))
print("Train/Test overlap:", len(train_ids & test_ids))
print("Val/Test overlap:", len(val_ids & test_ids))
```

## A3: Image preprocessing pipeline

**Choices made here, and why:**
- **Resize to 224x224** — this is the standard input size for many pretrained CNN backbones , so resizing to this size lets me use transfer learning without modifying the network's first layer.
- **Normalize pixel values to the 0-1 range** — raw PNG pixels are 0-255 integers. Neural networks train more stably on small, centred numeric ranges, so we divide by 255.
- **Light augmentation (rotation, zoom, horizontal flip) on training data only** — X-rays can appear at slightly different angles depending on how the scan was taken. Mild augmentation helps the model generalise rather than memorise exact pixel positions. 


```python
train_split = train_split.copy()
val_split = val_split.copy()
test_split = test_split.copy()

train_split["filename"] = train_split["id"].astype(str) + ".png"
val_split["filename"] = val_split["id"].astype(str) + ".png"
test_split["filename"] = test_split["id"].astype(str) + ".png"

train_split["age_bracket"] = train_split["age_bracket"].astype(str)
val_split["age_bracket"] = val_split["age_bracket"].astype(str)
test_split["age_bracket"] = test_split["age_bracket"].astype(str)

print(train_split[["id", "filename", "age_bracket"]].head())
```

### what this code does:

- **from tensorflow.keras.preprocessing.image import ImageDataGenerator**
  This imports Keras's tool for loading images in batches and applying transformations to them on the fly.

- **IMG_SIZE = (224, 224)** and **BATCH_SIZE = 32**
  These are settings used later when the images are loaded. `IMG_SIZE` resizes every image to 224x224 pixels so they all have the same shape, and 224x224 is the input size many pretrained models expect. `BATCH_SIZE` means the model sees 32 images at a time during training.

- **train_datagen = ImageDataGenerator(...)**
  This creates the generator for the training set, which applies augmentation. Each time an image is loaded, it is slightly altered, so the model sees a varied version of the data every epoch and is less likely to overfit.
  - `rescale=1.0 / 255` scales pixel values from the 0-255 range down to 0-1, which helps the model train more stably.
  - `rotation_range=10` randomly rotates each image by up to 10 degrees in either direction. This mimics small differences in how an X-ray is positioned.
  - `zoom_range=0.1` randomly zooms in or out by up to 10%. This mimics small differences in framing or distance.
  - `horizontal_flip=False` turns off left-right flipping. Left and right orientation can matter in medical images, so flipping could create X-rays that don't look realistic.

- **val_test_datagen = ImageDataGenerator(rescale=1.0 / 255)**
  This creates the generator for the validation and test sets. It only rescales the pixels and applies no augmentation. Evaluation should be done on the real, unaltered images so the scores reflect how the model performs on actual data. The pixel scaling must match the training generator so the model receives inputs in the same range.


```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = (224, 224)
BATCH_SIZE = 32 

train_datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    rotation_range=10,
    zoom_range=0.1,
    horizontal_flip=False,  
)

val_test_datagen = ImageDataGenerator(rescale=1.0 / 255)
```

```python
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_split,
    directory=train_images_path,
    x_col="filename",
    y_col="age_bracket",
    target_size=IMG_SIZE,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=True, 
    seed=42,
)

val_generator = val_test_datagen.flow_from_dataframe(
    dataframe=val_split,
    directory=train_images_path,  
    x_col="filename",
    y_col="age_bracket",
    target_size=IMG_SIZE,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=False, 
)

test_generator = val_test_datagen.flow_from_dataframe(
    dataframe=test_split,
    directory=train_images_path,
    x_col="filename",
    y_col="age_bracket",
    target_size=IMG_SIZE,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=False,
)

print("Class index mapping:", train_generator.class_indices)
```

```python
## A4: Model Architecture & Justification

**Architecture choice: transfer learning with MobileNetV2 as a frozen backbone, plus a custom classification head.**

**Why transfer learning rather than a custom CNN built from scratch:**
- We have ~7,566 training images spread across 4 classes — enough to fine-tune a classification head, but relatively small for training a deep CNN's convolutional filters entirely from scratch (which typically needs tens/hundreds of thousands of images to learn good low-level filters like edges and textures without overfitting).
- MobileNetV2 was pretrained on ImageNet (1.4 million general images). Its early layers already recognise general shapes, edges, and textures — useful even for X-rays, which are still images with edges and structures, just not photos of everyday objects. We reuse that general visual knowledge and only train new layers on top for our specific bone-age-bracket task.
- MobileNetV2 is lightweight compared to larger backbones (e.g. ResNet50, VGG16), which matters for training speed on a laptop without a dedicated GPU.

**What we rejected and why:** A custom CNN trained from scratch was considered, but was rejected due to the relatively small dataset size for this class count, which risks overfitting without the benefit of pretrained low-level features.

**Input shape:** (224, 224, 3) — 224x224 pixels, 3 colour channels. Note: X-rays are technically grayscale, but MobileNetV2 expects 3-channel input since it was trained on colour images; flow_from_dataframe automatically converts grayscale PNGs to 3-channel by default.

**Output shape:** (4,) — a probability distribution over our 4 age brackets (via softmax), e.g. [0.05, 0.10, 0.70, 0.15] meaning the model is 70% confident the image falls in the third bracket.
```

## A5: Training Process

**Loss function:** categorical crossentropy — the standard loss for multi-class classification where each image belongs to exactly one class.

**Optimizer:** Adam, with a small learning rate (0.0001) — Adam adapts its learning rate per parameter automatically and is a reliable default; we use a smaller-than-default learning rate since we're fine-tuning on top of pretrained weights, where large updates could destabilise the useful features already learned.

**Batch size:** 32 — a common default that balances training speed and memory usage.

**Epochs:** up to 20, with early stopping— rather than a fixed low number, we let the model train longer but automatically stop if it stops improving, to avoid both under- and over-training.

**Overfitting mitigation:** Dropout (0.3, already added above), plus **early stopping** (below), which halts training if validation loss stops improving for several epochs and restores the best-performing weights — directly addressing the risk of the model memorising the training set rather than generalising.


```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

base_model = MobileNetV2(
    weights="imagenet", include_top=False, input_shape=(224, 224, 3)
)
base_model.trainable = False


x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.3)(x)
outputs = Dense(4, activation="softmax")(x)

model = Model(inputs=base_model.input, outputs=outputs)


model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

early_stop = EarlyStopping(
    monitor="val_loss", patience=3, restore_best_weights=True
)

EPOCHS = 20

history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=[early_stop],
)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history.history["loss"], label="Training loss")
axes[0].plot(history.history["val_loss"], label="Validation loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].set_title("Training vs Validation Loss")
axes[0].legend()

axes[1].plot(history.history["accuracy"], label="Training accuracy")
axes[1].plot(history.history["val_accuracy"], label="Validation accuracy")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy")
axes[1].set_title("Training vs Validation Accuracy")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## A6: Evaluation & Results

We evaluate on the **test set** (2,523 images) — data the model has never seen during training or validation-based early stopping decisions.

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

test_generator.reset()
predictions = model.predict(test_generator, verbose=1)

predicted_classes = np.argmax(predictions, axis=1)

true_classes = test_generator.classes

class_labels = list(test_generator.class_indices.keys())

print(classification_report(true_classes, predicted_classes, target_names=class_labels))
```

```python
cm = confusion_matrix(true_classes, predicted_classes)

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_labels, yticklabels=class_labels, ax=ax)
ax.set_xlabel("Predicted bracket")
ax.set_ylabel("True bracket")
ax.set_title("Confusion Matrix — Test Set")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.image as mpimg
import os

test_split_reset = test_split.reset_index(drop=True)

correct_idx = None
incorrect_idx = None
for i in range(len(predicted_classes)):
    if predicted_classes[i] == true_classes[i] and correct_idx is None:
        correct_idx = i
    if predicted_classes[i] != true_classes[i] and incorrect_idx is None:
        incorrect_idx = i
    if correct_idx is not None and incorrect_idx is not None:
        break

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

for ax, idx, title in [(axes[0], correct_idx, "Correct prediction"),
                        (axes[1], incorrect_idx, "Incorrect prediction")]:
    img_path = os.path.join(train_images_path, test_split_reset.loc[idx, "filename"])
    img = mpimg.imread(img_path)
    ax.imshow(img, cmap="gray")
    true_label = class_labels[true_classes[idx]]
    pred_label = class_labels[predicted_classes[idx]]
    confidence = predictions[idx][predicted_classes[idx]] * 100
    ax.set_title(f"{title}\nTrue: {true_label}\nPredicted: {pred_label} ({confidence:.1f}%)")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## A8: Limitations, Ethics & Next Steps

**Data bias and generalisation limits:**
- The RSNA Bone Age dataset was collected from a specific set of US hospitals (Stanford, Colorado, UCLA), so the model may not generalise well to populations with different demographics, nutrition patterns, or ethnic backgrounds not well represented in this data, all of which can affect skeletal maturation.
- All images come from a limited set of scanning equipment/protocols; X-rays from different machines, exposure settings, or non-US clinical practices may look different enough to reduce model accuracy.
- Converting a continuous age (months) into 4 discrete brackets means the model's usefulness is bounded by those bracket widths — it cannot express fine-grained age estimates, and predictions near bracket boundaries are the least reliable (as seen in the confusion matrix in A6).

**Ethical considerations:**
- **False negatives** here predicting a younger/older bracket than reality, could lead to a missed or delayed referral for a genuine growth or endocrine concern.
- **False positives** could cause unnecessary parental anxiety or further unnecessary clinical investigation.
- Because consequences run in both directions and this model is trained on a relatively narrow, dataset-specific population, it must never be used as a standalone diagnostic decision-maker. It requires oversight from a qualified radiologist or paediatrician who can weigh the prediction against the full clinical picture (patient history, physical exam, other tests).

**What this project cannot claim:** This is a personal project demonstrating an ML pipeline, not a validated clinical tool. It has not been tested against a clinically diverse population, has not undergone any regulatory review, and its accuracy figures reflect performance on a single dataset's own held-out split, not real-world deployment conditions.

**What would be needed before real clinical use:** Validation on independent, demographically diverse datasets. Review and sign-off from clinical experts. Clear guardrails ensuring it's used only as a decision-support aid alongside, not instead of, a qualified clinician.
