### Activating the Vision 

* **`import cv2`**
  After downloading OpenCV with pip, this line actually opens the toolbox and brings it into the script. It unlocks advanced functions like converting color spaces (like RGB to HSV), filtering out noise, and finding borders in images.


```python
image = mpimg.imread('Betty.jpg')
print('Image dimension:', image.shape)
plt.imshow(image)
plt.show()
```

### Loading and Showing the Image

* **`image = mpimg.imread('Profilepic.png')`**: **Opening the photo.** 
  This line looks inside your folder for a file named `'Profilepic.png'`, reads it, and transforms it into a grid of numbers for Python to work with.
* **`print('Image dimension:', image.shape)`**: **Checking the size.** 
  This prints out the exact dimensions of your image (Height, Width, and the 3 color channels: Red, Green, and Blue).
* **`plt.imshow(image)`**: **Setting up the canvas.** 
  This prepares the number grid to be drawn as a visual picture on a graph.
* **`plt.show()`**: **Displaying the photo.** 
  This tells your screen to finally display the visual picture so you can see it in your notebook.


```python
gray_image=cv2.cvtColor(image,cv2.COLOR_RGB2GRAY)
plt.imshow(gray_image, cmap='gray')        
```

### Converting to Black and White

* **`gray_image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)`**: **Stripping the color.** 
  This line takes your colorful image and uses OpenCV to remove all the color information, turning it into a clean, black-and-white (grayscale) picture.
* **`plt.imshow(gray_image, cmap='gray')`**: **Displaying true gray.** 
  This prepares the black-and-white image to be shown on your screen. The `cmap='gray'` part is super important—it tells Python to use actual gray shades, instead of displaying the image with weird default purple and yellow tints.


```python
r=image[:,:,0]
g=image[:,:,1]
b=image[:,:,2]

f, (ax1, ax2, ax3)=plt.subplots(1,3,figsize=(20,10))
ax1.set_title('R channel')
ax1.imshow(r, cmap='gray')
ax2.set_title('G channel')
ax2.imshow(g, cmap='gray')
ax3.set_title('B channel')
ax3.imshow(b, cmap='gray')
```

### Splitting the Image into Red, Green, and Blue Channels

* **`r = image[:, :, 0]` (and `g`, `b`)**: **Slicing the color layers.** 
  Digital images are like sandwiches made of three color layers: Red (0), Green (1), and Blue (2). The `:` symbols mean "take all rows and columns," while the final number pulls out just that single color layer.
* **`f, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 10))`**: **Creating a 3-panel display window.** 
  This sets up a clean, wide row (`1` row, `3` columns) of empty boxes so we can look at all three color layers side by side. `figsize=(20, 10)` makes the display window extra large and easy to see.
* **`ax1.set_title('R channel')`**: **Labeling the box.** 
  This puts a helpful name above the first window box so you know exactly which color channel you are looking at.
* **`ax1.imshow(r, cmap='gray')`**: **Showing the layer intensity.** 
  This draws the single color channel into its box. Even though it is a color channel, we show it in gray (`cmap='gray'`). In this view, **bright white spots** mean that specific color is very strong in that part of the photo, while **dark black spots** mean that color is missing.


```python
hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)

saturation = hsv[:, :, 1]

plt.figure(figsize=(6, 6))

plt.imshow(saturation, cmap='gray')
plt.title('Saturation Channel')
plt.axis('off')
plt.show()
```

```python
hsv=cv2.cvtColor(image,cv2.COLOR_RGB2HSV)
h=hsv[:,:,0]
s=hsv[:,:,1]
v=hsv[:,:,2]

f, (ax1, ax2, ax3)=plt.subplots(1,3,figsize=(20,10))
ax1.set_title('Hue')
ax1.imshow(h, cmap='gray')

ax2.set_title('Saturation')
ax2.imshow(s, cmap='gray')

ax3.set_title('Value')
ax3.imshow(v, cmap='gray')
```
